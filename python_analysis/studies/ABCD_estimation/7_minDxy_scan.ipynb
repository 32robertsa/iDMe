```python
cut = 'cut10'
```

```python
m1s = [5, 10, 20, 40, 50, 70, 90, 100]
deltas = [0.1, 0.2]
ctaus = [1, 10, 100]
```

```python
range_scan = list(np.arange(0.001, 0.1, 0.001))

sig_As = {}
sig_ABs = {}
sig_effs = {}
```

```python
range_scan
```

```python
samp_df = si[(si.m1 == 10) & (si.delta == 0.1) & (si.ctau == 1)]
samp = samp_df.name[0]

sig_histo['sel_min_dxy_refit'][{"samp":samp, "cut": cut}][99:]
```

```python
for m1 in m1s:
    for delta in deltas:
        for ctau in ctaus:
            samp_df = si[(si.m1 == m1) & (si.delta == delta) & (si.ctau == ctau)]
            samp = samp_df.name[0]

            sig_As[samp] = []
            sig_ABs[samp] = []
            sig_effs[samp] = []

            print(f'\n m1 = {m1}, delta = {delta}, ctau = {ctau}')

            for idx in range(len(range_scan)):
                # Yields
                sig_A = sig_histo['sel_min_dxy_refit'][{"samp":samp, "cut": cut}][idx:].sum().value
                sig_AB = sig_histo['sel_min_dxy_refit'][{"samp":samp, "cut": cut}][:].sum().value
                
                print(f'dxt cut ({range_scan[idx]:.4f}) efficiency = {sig_A:.1f}/{sig_AB:.1f} = {sig_A/sig_AB:.4f}')
    
                sig_As[samp].append(sig_A)
                sig_ABs[samp].append(sig_AB)
                sig_effs[samp].append(sig_A/sig_AB)
```

```python
sig_cut_idx_0p9eff = {}
```

```python
for sig_point, eff_list in sig_effs.items():
    # Find the eff value closest to 90%
    delta_eff = np.abs(np.array(eff_list) - 0.9)
    print(sig_point)
    print(f'{eff_list[np.argmin(delta_eff)]:.2f}')

    sig_cut_idx_0p9eff[sig_point] = np.argmin(delta_eff)
```

```python
sig_cut_idx_0p9eff
```

```python
bkg_As = []
bkg_ABs = []
bkg_effs = []
```

```python
for idx in range(len(range_scan)):
    bkg_A = 0
    bkg_AB = 0
    
    for i in range(len(bkg_samp)):
        bkg_A += bkg_histo['sel_min_dxy_refit'][{"samp":bkg_samp[i], "cut": cut}][idx:].sum().value
        bkg_AB += bkg_histo['sel_min_dxy_refit'][{"samp":bkg_samp[i], "cut": cut}][:].sum().value
    
    print(f'dxy cut ({range_scan[idx]:.4f}) efficiency = {bkg_A:.1f}/{bkg_AB:.1f} = {bkg_A/bkg_AB:.4f}')

    bkg_As.append(bkg_A)
    bkg_ABs.append(bkg_AB)
    bkg_effs.append(bkg_A/bkg_AB)
```

```python
s_over_b_max_idx = {}
s_over_b_max = {}
```

```python
for sig_point, yield_list in sig_As.items():
    s_over_b = np.array(yield_list)/np.sqrt(bkg_As)

    print(sig_point)
    print(f'{np.argmax(s_over_b)}, {s_over_b[np.argmax(s_over_b)]:.3f}')

    s_over_b_max_idx[sig_point] = np.argmax(s_over_b)
    s_over_b_max[sig_point] = s_over_b[np.argmax(s_over_b)]
```

```python
s_over_b_max_idx
```
