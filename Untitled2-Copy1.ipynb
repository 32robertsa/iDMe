```python
genelevxy1000 = tree1000["GenEle_vxy"].array()

plt.hist(genelevxy1000, bins=100,histtype="step",color = 'red', linewidth=1.5, label="gen vxy 1000")

plt.yscale("log")
plt.xlabel("GenEle_vxy [cm]")
plt.legend()
plt.show()
```

```python
pdgs1000 = tree1000["GenPart_ID"].array()
mask1000 = pdgs1000==1000023
```

```python
energy1000 = tree1000["GenPart_e"].array()
energychi21000=energy1000[mask1000]

mass1000 = tree1000["GenPart_mass"].array()
mchi21000 = mass1000[mask1000]
```

```python
gamma1000= energychi21000/mchi21000

#plt.hist(gamma)
plt.hist(gamma1000, bins=100, histtype="step",color = 'red',linewidth=1.5, label="gen vxy 1000")
plt.xlabel("Gamma factor= E/mchi2")
plt.legend()
plt.show()
```

```python
ctau1000= genelevxy1000/gamma1000

y,x, _=plt.hist(ctau1000, bins=100, histtype="step",linewidth=1.5, range=(0,1000),log=True, color = 'red', label="ctau0= 1000 mm sample")

print (y)  #count per bin
print (x)

ct= (x[:-1] + x[1:])/2   #the ct values (in cm)
print (ct)
```

```python
from scipy.optimize import curve_fit

def expo(ct, A, ctau0):
    return (A * np.exp(-ct/ctau0))
```

```python
p0 = [10000,100]   # initial guess: [A, tau]

params, cov = curve_fit(expo,ct, y, p0=p0)

A_fit, ctau0_fit = params
print("Fitted ctau0 =", ctau0_fit)  #this is for a 100 cm sample
```

```python
plt.hist(ctau1000, bins=100, range=(0,1000),
         histtype="step", linewidth=1.5, 
         color='red', label="ctau0 = 100 cm sample")

plt.plot(ct, expo(ct, A_fit, tau_fit),
         color='black', label=f"Fit: ctau0 = {ctau0_fit:.1f}")

plt.xlabel("ctau (cm)")
plt.legend()
plt.savefig("ctau_histogram.png", dpi=300)
plt.show()
```
