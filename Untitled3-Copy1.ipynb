```python
%load_ext autoreload
%autoreload 2
import numpy as np
import uproot
import awkward as ak
import matplotlib.pyplot as plt
from coffea import util
from coffea.processor import accumulate
import os
import re
import pandas as pd
import hist
import matplotlib as mpl
import mplhep
import sys
```

```python
file10 = uproot.open("Testctau10.root")
dir10= file10["ntuples"]    
tree10 = dir10["outT"]

file100 = uproot.open("Testctau100.root")
dir100= file100["ntuples"]    
tree100 = dir100["outT"]
```

```python
genelevxy10 = tree10["GenEle_vxy"].array()
genelevxy100 = tree100["GenEle_vxy"].array()

plt.hist(genelevxy10, bins=100,histtype="step",color = 'blue',linewidth=1.5, label="gen vxy 10")
plt.hist(genelevxy100, bins=100,histtype="step",color = 'red', linewidth=1.5, label="gen vxy 100")

plt.yscale("log")
plt.xlabel("GenEle_vxy [cm]")
plt.legend()
plt.show()
```

```python
pdgs10 = tree10["GenPart_ID"].array()
mask10 = pdgs10==1000023

pdgs100 = tree100["GenPart_ID"].array()
mask100 = pdgs100==1000023
```

```python
energy10 = tree10["GenPart_e"].array()
energychi210=energy10[mask10]

mass10 = tree10["GenPart_mass"].array()
mchi210 = mass10[mask10]

energy100 = tree100["GenPart_e"].array()
energychi2100=energy100[mask100]

mass100 = tree100["GenPart_mass"].array()
mchi2100 = mass100[mask100]
```

```python
gamma10= energychi210/mchi210
gamma100= energychi2100/mchi2100

#plt.hist(gamma)

plt.hist(gamma10, bins=100, histtype="step",color = 'blue',linewidth=1.5, label="gen vxy 10")
plt.hist(gamma100, bins=100, histtype="step",color = 'red',linewidth=1.5, label="gen vxy 100")
plt.xlabel("Gamma factor= E/mchi2")
plt.legend()
plt.show()
```

```python
ctau10= genelevxy10/gamma10
ctau100= genelevxy100/gamma100

plt.hist(ctau10, bins=100, histtype="step",range=(0,20),linewidth=1.5, log=True, color = 'blue', label="gen vxy 10")
plt.hist(ctau100, bins=100, histtype="step",range=(0,20),linewidth=1.5, log=True, color = 'red', label="gen vxy 100")

plt.xlabel("ctau (in cm)")
plt.savefig("ctau_histogram.png", dpi=300)
plt.legend()
plt.show()
```
